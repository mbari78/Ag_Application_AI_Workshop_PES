```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# Load your data
# Uncomment if you need to load the data
# data = pd.read_csv('your_file_path.csv')
data1 = data1

# Select columns for correlation matrix
columns_of_interest = ['Yld_ton_ha', 'DG_Pan_Count', 'B_Box_Pan_Area_cm2', 'Lab_Pan_Area_cm2', 'Mask_Pan_Area_cm2', 'Predict_Pan_Size_Pixel', 'Pan_Seed_Count', 'Reg_Seed_Count', 'Spr_Seed_Count', 'Microscope_S_Area_cm2', 'DG_Seed_Area_mm2',
       'SKCS_Seed_dia_mm']
data_subset = data1[columns_of_interest]

# Calculate the correlation matrix
correlation_matrix = data_subset.corr()

# Initialize an empty significance matrix
significance_matrix = np.zeros(correlation_matrix.shape)

# Calculate p-values and add significance asterisks
for i in range(len(columns_of_interest)):
    for j in range(len(columns_of_interest)):
        if i != j:
            _, p_value = pearsonr(data_subset[columns_of_interest[i]], data_subset[columns_of_interest[j]])
            significance_matrix[i, j] = p_value
        else:
            significance_matrix[i, j] = np.nan  # Diagonal elements set to NaN

# Define a function to add asterisks based on p-value thresholds
def significance_level(p):
    if p < 0.001:
        return "***"  # Very significant
    elif p < 0.01:
        return "**"  # Significant
    elif p < 0.05:
        return "*"  # Slightly significant
    else:
        return ""  # Not significant

# Create annotations combining correlation values with significance levels
annotations = correlation_matrix.round(2).astype(str)
for i in range(len(columns_of_interest)):
    for j in range(len(columns_of_interest)):
        if i != j:
            # Add asterisks next to the correlation values
            annotations.iloc[i, j] += significance_level(significance_matrix[i, j])

# Create a mask for the upper triangle
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

# Set up the matplotlib figure
plt.figure(figsize=(10, 8))

# Draw the heatmap with annotated values and asterisks
sns.heatmap(correlation_matrix, mask=mask, annot=annotations, fmt="", cmap='coolwarm',
            vmax=1, vmin=-1, center=0, square=True, linewidths=.5, cbar_kws={"shrink": .75})

# Set title and display the plot
plt.title("Correlation among yield and digitally obtained yield attributing features")
plt.show()
```

## Non-Collinear Feature

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler

# Step 0: Define X and y
X = df.drop(columns=['Plot_Pedigree', 'Plot', 'Pedigree', 'Range', 'Row', 'Pass', 'Rep', 'Yld_ton_ha', 'GT_Pan_Count', 'GT_Pan_Size_cm2', 'Mach_Seed_Count'])
y = df['Yld_ton_ha']

# Step 1: Scale features (optional but good for consistent importance estimation)
scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)

# Step 2: Fit Random Forest
rf = RandomForestRegressor(random_state=42)
rf.fit(X_scaled, y)

# Step 3: Get feature importances
importances = pd.Series(rf.feature_importances_, index=X.columns)
importances_sorted = importances.sort_values(ascending=False)

# Step 4: Remove highly correlated features
def remove_correlated_features(df, feature_list, threshold=0.85):
    selected = []
    corr_matrix = df[feature_list].corr().abs()

    for feature in feature_list:
        keep = True
        for sel in selected:
            if corr_matrix.loc[feature, sel] >= threshold:
                keep = False
                break
        if keep:
            selected.append(feature)
    return selected

# Step 5: Apply collinearity filter to sorted features
top_candidates = importances_sorted.index.tolist()
non_collinear_top_features = remove_correlated_features(X_scaled, top_candidates)

# Step 6: Get top N features
top_features_final = non_collinear_top_features[:6]

# Output
print("Top 6 non-collinear features for predicting Yld_ton_ha:")
print(top_features_final)
```
